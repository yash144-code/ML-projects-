```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input,Dense,SimpleRNN
import random

np.random.seed(42)
tf.random.set_seed(42)

numbers = np.arange(1,101,dtype = np.float32)
numbers = numbers/100.0

X =[]
y=[]

for i in range(len(numbers)-3):
  X.append(numbers[i:i+3])
  y.append(numbers[i+3])

X = np.array(X, dtype =np.float32)
y= np.array(y,dtype = np.float32)

X =X.reshape(X.shape[0],X.shape[1],1)

# Split the data in time order
X_train = X[:77]
y_train = y[:77]

X_test = X[77:]
y_test = y[77:]

print("Training input shape:", X_train.shape)
print("Training target shape:", y_train.shape)

print("Testing input shape:", X_test.shape)
print("Testing target shape:", y_test.shape)

np.random.seed(42)
tf.random.set_seed(42)

model = Sequential([
    Input(shape=(3, 1)),
    SimpleRNN(4),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

history = model.fit(
    X_train,
    y_train,
    epochs=500,
    verbose=0
)

print("Final training loss:", history.history["loss"][-1])
```

```python
test_loss = model.evaluate(X_test,y_test,verbose=0)

print("Training loss:", history.history["loss"][-1])
print("Testing loss:", test_loss)
```

```python
import random
import numpy as np
import tensorflow as tf
from sklearn.metrics import mean_absolute_error

random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

tf.config.experimental.enable_op_determinism()
model = Sequential([
    Input(shape=(3, 1)),
    SimpleRNN(4),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)
history = model.fit(
    X_train,
    y_train,
    epochs=500,
    verbose=0
)

test_loss = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Training loss:", history.history["loss"][-1])
print("Testing loss:", test_loss)

predictions = model.predict(X_test, verbose=0).flatten()

actual = y_test * 100
predicted = predictions * 100

mae = mean_absolute_error(actual, predicted)

print("Test MAE:", mae)
```

```python
for i in range(len(X_test)):
    sequence = (X_test[i].flatten() * 100).astype(int)

    print(
        f"Input: {sequence.tolist()} | "
        f"Actual: {actual[i]:.2f} | "
        f"Predicted: {predicted[i]:.2f} | "
        f"Error: {abs(actual[i] - predicted[i]):.2f}"
    )

errors = np.abs(actual - predicted)

print("Sum of absolute errors:", np.sum(errors))
print("Number of test examples:", len(errors))
print("Manual MAE:", np.mean(errors))
```

```python
# Create numbers normalized by 1000
numbers_1000 = np.arange(
    1, 101, dtype=np.float32
) / 1000.0

X_1000 = []
y_1000 = []

for i in range(len(numbers_1000) - 3):
    X_1000.append(numbers_1000[i:i+3])
    y_1000.append(numbers_1000[i+3])

X_1000 = np.array(X_1000, dtype=np.float32)
y_1000 = np.array(y_1000, dtype=np.float32)

X_1000 = X_1000.reshape(
    X_1000.shape[0], X_1000.shape[1], 1
)

# Preserve time order
X_train_1000 = X_1000[:77]
y_train_1000 = y_1000[:77]

X_test_1000 = X_1000[77:]
y_test_1000 = y_1000[77:]

print("X_train:", X_train_1000.shape)
print("X_test:", X_test_1000.shape)
```

```python
# Reset random seeds
import random
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# Build a fresh model
model_1000 = Sequential([
    Input(shape=(3, 1)),
    SimpleRNN(64),
    Dense(1)
])

model_1000.compile(
    optimizer="adam",
    loss="mse"
)

# Train only on training data
history_1000 = model_1000.fit(
    X_train_1000,
    y_train_1000,
    epochs=500,
    verbose=0
)

# Evaluate on the reserved test data
test_loss_1000 = model_1000.evaluate(
    X_test_1000,
    y_test_1000,
    verbose=0
)

print("Training loss:", history_1000.history["loss"][-1])
print("Testing loss:", test_loss_1000)
```

```python
from sklearn.metrics import mean_absolute_error

predictions_1000 = model_1000.predict(
    X_test_1000,
    verbose=0
).flatten()

# Convert to the original number scale
actual_1000 = y_test_1000 * 1000
predicted_1000 = predictions_1000 * 1000

mae_1000 = mean_absolute_error(
    actual_1000,
    predicted_1000
)

print("Test MAE (original scale):", mae_1000)


test_sequence_1000 = np.array(
    [101, 102, 103],
    dtype=np.float32
) / 1000.0

test_sequence_1000 = test_sequence_1000.reshape(1, 3, 1)

prediction_1000 = model_1000.predict(
    test_sequence_1000,
    verbose=0
)

print("Expected:", 104)
print("Predicted:", prediction_1000[0][0] * 1000)
```

```python
test_sequences = [
    [50, 51, 52],          # Within the original data range
    [75, 76, 77],          # Within the original data range
    [101, 102, 103],       # Just beyond training range
    [150, 151, 152],       # Further beyond
    [200, 201, 202],       # Further still
    [500, 501, 502],       # Much further beyond
    [1000, 1001, 1002]     # Far beyond
]

for sequence in test_sequences:
    x_new = np.array(
        sequence, dtype=np.float32
    ) / 1000.0

    x_new = x_new.reshape(1, 3, 1)

    prediction = model_1000.predict(
        x_new, verbose=0
    )[0][0] * 1000.0

    expected = sequence[-1] + 1

    print(
        f"Input: {sequence} | "
        f"Expected: {expected} | "
        f"Predicted: {prediction:.3f} | "
        f"Error: {abs(expected - prediction):.3f}"
    )
```
